```python
import pandas as pd

# 1. CSV 파일 읽기
file1 = '로우법안데이터/data20_반영여부.csv'  # 1번 파일 이름
file2_before = '의원데이터/전체데이터/제20대_법안별_전체데이터_node.csv'  # 이전 노드 파일
file2_after = '의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이후_node.csv'  # 이후 노드 파일

df1 = pd.read_csv(file1)

# 2. 제안일 기준으로 데이터 나누기
df1['제안일'] = pd.to_datetime(df1['제안일'])  # 제안일 열을 datetime 형식으로 변환
cutoff_date = pd.Timestamp('2018-02-13')
#aa = pd.Timestamp('2018-02-13')

df_before = df1[df1['제안일'] < cutoff_date]
#df_before = df_before[df_before['제안일'] >= aa]
df_after = df1[df1['제안일'] >= cutoff_date]

# 3. 각 시기의 노드 파일 읽기
df2_before = pd.read_csv(file2_before)
df2_after = pd.read_csv(file2_after)

# 4. 대표발의자열과 공동발의자열 합치기
def process_parties(df, node_df):
    merged_names = df['대표발의자'].fillna('') + ',' + df['공동발의자'].fillna('')
    merged_names = merged_names.str.strip(',')  # 불필요한 ',' 제거

    # 이름에 따라 노드 파일에서 Party 매칭
    party_mapping = node_df.set_index('Id')['Party'].to_dict()

    def map_party(name_list):
        parties = [party_mapping.get(name.strip(), 'Unknown') for name in name_list.split(',')]
        return list(filter(lambda x: x != 'Unknown', parties))

    df = df.copy()  # 원본 DataFrame 수정 방지
    df.loc[:, 'Parties'] = merged_names.apply(map_party)
    return df

# 처리된 데이터
processed_before = process_parties(df_before, df2_before)
processed_after = process_parties(df_after, df2_after)

# 5. 노드 수가 많은 상위 정당 추출 함수
def count_parties(parties):
    party_count = {}
    for party_list in parties:
        for party in set(party_list):
            if party != 'Unknown':
                party_count[party] = party_count.get(party, 0) + 1
    return party_count

def get_top_parties(processed_df, top_n=3):
    party_counts = count_parties(processed_df['Parties'])
    sorted_parties = sorted(party_counts.items(), key=lambda x: x[1], reverse=True)
    return [p[0] for p in sorted_parties[:top_n]]

# 6. 3번째 정당 법안에서 협력 비율 계산 함수
def calculate_fixed_party_cooperation(processed_df,pa):
    first_second_parties = {'더불어민주당', '자유한국당'}  # 상위 2개 정당 고정
    third_party = pa  # 3번째 정당 고정

    def is_third_party_representative(parties):
        return third_party in parties[:1]  # 대표발의자가 3번째 정당인지 확인

    def is_first_second_cooperation(parties):
        involved_parties = set(parties)
        return len(first_second_parties & involved_parties) == 2  # 1, 2위 정당이 모두 포함되었는지 확인

    filtered_df = processed_df[processed_df['Parties'].apply(is_third_party_representative)]
    filtered_df = filtered_df.copy()
    filtered_df['Is_First_Second_Cooperation'] = filtered_df['Parties'].apply(is_first_second_cooperation)

    total_laws = len(filtered_df)
    cooperative_laws = filtered_df['Is_First_Second_Cooperation'].sum()

    return cooperative_laws / total_laws if total_laws > 0 else 0.0

# 7. 협력 비율 계산
third_party_before_ratio = calculate_fixed_party_cooperation(processed_before,"국민의당")
third_party_after_ratio = calculate_fixed_party_cooperation(processed_after,"바른미래당")

# 결과 출력
print(f"2018-02-13 이전 국민의당 대표발의 법안에서 더불어민주당과 자유한국당 협력 비율: {third_party_before_ratio:.2%}")
print(f"2018-02-13 이후 바른미래당 대표발의 법안에서 더불어민주당과 자유한국당 협력 비율: {third_party_after_ratio:.2%}")
```

```python
import pandas as pd
import networkx as nx
from collections import Counter

# 1. 엣지와 노드 CSV 파일 목록 정의
edge_files = [
    '의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이전_edge.csv',
    '의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이후_edge.csv',
    '의원데이터/전체데이터/제20대_법안별_전체데이터_edge.csv',
    '의원데이터/전체데이터/제21대_법안별_전체데이터_edge.csv'
]

node_files = [
    '의원데이터/전체데이터/제20대_법안별_전체데이터_node.csv',
    '의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이후_node.csv',
    '의원데이터/전체데이터/제20대_법안별_전체데이터_node.csv',
    '의원데이터/전체데이터/제21대_법안별_전체데이터_node.csv'
]

## 2. 정당별 혼합성 지수를 계산하는 함수 정의
def calculate_party_assortativity(graph, party, attribute='Party', weight='weight'):
    """
    특정 정당에 대한 혼합성 지수를 계산합니다.

    Parameters:
        graph (nx.DiGraph): 네트워크 그래프
        party (str): 정당 이름
        attribute (str): 노드 속성 키 (예: 'Party')
        weight (str): 엣지 가중치 속성 키 (기본값: 'weight')

    Returns:
        float: 특정 정당에 대한 혼합성 지수
    """
    edges = graph.edges(data=True)
    weighted_sum = 0
    total_weight = 0

    for u, v, data in edges:
        attr_u = graph.nodes[u].get(attribute, None)
        attr_v = graph.nodes[v].get(attribute, None)
        w = data.get(weight, 1)  # 기본 가중치 1

        if attr_u == party or attr_v == party:
            weighted_sum += w * (attr_u == attr_v)
            total_weight += w

    assortativity = weighted_sum / total_weight if total_weight > 0 else 0
    return assortativity

# 3. 각 네트워크에 대해 처리
for edge_file, node_file in zip(edge_files, node_files):
    # 3.1 엣지 데이터 읽기
    edges_df = pd.read_csv(edge_file)

    # 3.2 노드 데이터 읽기
    nodes_df = pd.read_csv(node_file)

    # 3.3 방향성 그래프 생성
    G = nx.DiGraph()

    # 3.4 노드 추가
    for index, row in nodes_df.iterrows():
        G.add_node(row['Id'],
                   Label=row['Label'],
                   Party=row['Party'],
                   Reselection=row['Reselection'],
                   District_Type=row['District_Type'],
                   Gender=row['Gender'],
                   Role=row['Role'])

    # 3.5 엣지 추가 (중복 엣지 가중치 합산)
    for index, row in edges_df.iterrows():
        if G.has_edge(row['Source'], row['Target']):
            # 이미 존재하는 엣지의 가중치에 추가
            G[row['Source']][row['Target']]['weight'] += 1
        else:
            # 새 엣지를 추가
            G.add_edge(row['Source'], row['Target'], weight=1)

    # 4. 각 정당별 혼합성 지수 계산
    parties = set(nx.get_node_attributes(G, 'Party').values())
    for party in parties:
        # 특정 정당에 대한 혼합성 지수 계산
        assortativity_index = calculate_party_assortativity(G, party)
        
        # 결과 출력
        print(f"엣지 파일 '{edge_file}'의 정당 '{party}' 혼합성 지수: {assortativity_index:.4f}")
```
