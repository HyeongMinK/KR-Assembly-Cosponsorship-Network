```python

import pandas as pd

# 1. CSV 파일 읽기
edges_df = pd.read_csv('의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이전_edge.csv')

# 2. Source와 Target 값 정렬
def sort_source_target(row):
    # Source와 Target을 정렬하여 작은 값을 Source에, 큰 값을 Target에 둠
    if row['Source'] > row['Target']:
        # Source와 Target 교환
        row['Source'], row['Target'] = row['Target'], row['Source']
        # 관련 속성도 교환
        row['Source_Party'], row['Target_Party'] = row['Target_Party'], row['Source_Party']
        row['Source_Reselection'], row['Target_Reselection'] = row['Target_Reselection'], row['Source_Reselection']
        row['Source_District_Type'], row['Target_District_Type'] = row['Target_District_Type'], row['Source_District_Type']
        row['Source_Gender'], row['Target_Gender'] = row['Target_Gender'], row['Source_Gender']
        row['Source_Role'], row['Target_Role'] = row['Target_Role'], row['Source_Role']
    return row

# 3. 각 행을 정렬
edges_df = edges_df.apply(sort_source_target, axis=1)

# 4. 정리된 데이터를 CSV로 저장
edges_df.to_csv('의원데이터/전체데이터/제20대_법안별_전체데이터__2018-02-13_이전_edge_무방향.csv', index=False)

print("Source와 Target 정렬이 완료되었습니다: edges_sorted.csv")
```
